```python
#       Q-1). write a program that takes two numbers (a = 15, b = 4) & prints their:
# (1- Sum) 
# (2- Difference)
# (3- Product)
# (4- Division)
# (5- Floor Division)
# (6- Modulus) 
```

```python
a = 15
b = 4

Sum_of_Number = a + b
Difference_of_Number = a - b
Product_of_Number = a * b
Division_of_Number = a / b
Floor_Division_of_Number = a // b
Modulus_of_Number = a % b
```

```python
print("Sum_of_Number is: ", Sum_of_Number)
print("Difference_of_Number is: ", Difference_of_Number)
print("Product_of_Number is: ", Product_of_Number)
print("Division_of_Number is: ", Division_of_Number)
print("Floor_Division_of_Numbe is: ", Floor_Division_of_Number)
print("Modulus_of_Number is: ", Modulus_of_Number)
```
