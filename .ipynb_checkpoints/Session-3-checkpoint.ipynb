# ***List***

```python
# In Python, a list is a built-in data structure used to store an ordered collection of items in a single variable.
# Lists are one of the most versatile and commonly used data types in Python.

# (In Python, a list is an ordered, mutable collection of items enclosed in square brackets []).index no starting is 0.
    
# Ordered:=           The items maintain a defined order. When you add new items, they are placed at the end of the list (by default).
# Mutable (Changeable):= You can modify, add, or remove items after the list has been created.
# Allow Duplicates:=  Because items are indexed, a list can contain multiple elements with the exact same value.
# Heterogeneous:=     A single list can hold different data types simultaneously (e.g., strings, integers, floats, and even other lists).
```

# ***Slicing***

```python
# Slicing (Ek se zyada items ya range nikalna)

# Slicing ke zariye aap sequence ka ek tukda kaat kar nikal sakte hain. Iska formula yeh hota hai:
# sequence[start : stop : step]
# • start: Jahan se slice shuru karna hai (yeh shaamil hota hai).
# • stop: Jahan tak slice karna hai (yeh index shaamil nahi hota, is se ek pehle tak hota hai).
# • step (Optional): Kitne gaps chhor kar chalna hai (by default yeh 1 hota hai).
```

```python
alphabatic_foods = ["Apple", "Banana", "Coconut ", "Date", "Elderberry", "Fig", "graps", "Honeydew_Melon", "Ice Apple"]

alphabatic_foods
```

```python
alphabatic_foods[-1]
```

```python
alphabatic_foods[1:] # slice 

# Docstring:  
# Built-in mutable sequence.

# If no argument is given, the constructor creates a new empty list.
# The argument must be an iterable if specified.
```

```python
alphabatic_foods[0:4]

# ek exclude krta ha 0 to 4 =5
```

```python
print( alphabatic_foods[ -4: -1 ] ) # negative ma bh 1 exclude krdia
```

#### ***Mannual Approach = item index change by slicing concept*** 

```python
alphabatic_foods
```

```python
alphabatic_foods[4:7] = "Blue_Barry", "papita", "waterMellon"
```

```python
alphabatic_foods
```
