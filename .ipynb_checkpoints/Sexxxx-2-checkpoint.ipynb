#### ***# Maths operations " BODMAS***

```python
# Maths operations " BODMAS = bracket of divisions, Multiplication, Addition & then subtraction 

# +  	Addition	Numbers ko add karta hai
# -	   Subtraction	Numbers ko subtract karta hai
# *	   Multiplication	Numbers ko multiply karta hai
# /	   Division Decimal answer deta hai
# //   Floor Division	Decimal ko floor karta hai
# **   Exponentiation	Power calculate karta hai
# %	   Modulo	Remainder deta hai
```

```python
print(4 + 5 ,'Addition')    # Addition
print(4 - 5 , 'Subtraction')    # Subtraction
print(4 * 5, 'Multiplication')    # Multiplication / product
print(4 / 5, 'Division floating answer')    # Division (returns float)
print(4 // 5, 'Floor division')   # Floor division
print(4 ** 5, 'Exponentiation (power)')   # Exponentiation (power)
print(10 % 3, 'Modulus remainder')   # Modulo (returns remainder)

print(int(5.87)) # Convert float into integer
```
